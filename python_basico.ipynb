## 1. Sintaxis Básica y Variables

### Diferencias clave con MATLAB:
- **No se necesita punto y coma** (`;`) al final de las líneas
- **Indentación obligatoria**: Python usa espacios/tabulaciones para definir bloques de código
- **Indexación empieza en 0** (MATLAB empieza en 1)

````python
# Operaciones aritméticas básicas
suma = 10 + 5
resta = 10 - 5
multiplicacion = 10 * 5
division = 10 / 3        # División con decimales
division_entera = 10 // 3  # División entera (equivalente a floor en MATLAB)
potencia = 2**3          # MATLAB: 2^3
modulo = 10 % 3

print(f"División: {division:.2f}")  # .2f formatea a 2 decimales
print(f"División entera: {division_entera}")
print(f"Potencia: {potencia}")
````

## 2. Listas y Tuplas

### Listas
- Equivalente a arrays/celdas en MATLAB
- Se definen con corchetes `[]`
- Son **mutables** (se pueden modificar)

````python
# Crear listas
ciudades = ['NYC', 'LA', 'Chicago', 'Houston']
numeros = [1, 2, 3, 4, 5]
mixta = [1, 'dos', 3.0, True]  # Python permite tipos mixtos

print("Primera ciudad:", ciudades[0])      # Índice 0 = primer elemento
print("Última ciudad:", ciudades[-1])      # Índice -1 = último elemento
print("Segunda y tercera:", ciudades[1:3]) # Slicing [inicio:fin] (fin no incluido)
````

````python
# Operaciones con listas
almacenes = ['Harlingen', 'Memphis', 'Ashland']

# Añadir elementos
almacenes.append('Dallas')      # Añade al final
print("Después de append:", almacenes)

# Longitud de la lista
n = len(almacenes)              # MATLAB: length()
print(f"Número de almacenes: {n}")

# Verificar si un elemento existe
if 'Memphis' in almacenes:
    print("Memphis está en la lista")
````

### Tuplas
- Se definen con paréntesis `()`
- Son **inmutables** (no se pueden modificar después de crear)
- **Muy importantes en Pyomo**: se usan para indexar variables con múltiples índices

````python
# Tuplas: útiles para pares de índices en Pyomo
par = ('Harlingen', 'NYC')  # Tupla de almacén-ciudad
coordenada = (10, 20, 30)

print("Primer elemento del par:", par[0])
print("Segundo elemento del par:", par[1])

# Las tuplas se usan mucho en Pyomo para variables indexadas:
# x[('Harlingen', 'NYC')] o x['Harlingen', 'NYC']
````

## 3. Diccionarios

**¡MUY IMPORTANTE PARA PYOMO!**

Los diccionarios almacenan pares **key-value**. Son equivalentes a `struct` en MATLAB.

- Se definen con llaves `{}`
- Formato: `{key1: value1, key2: value2}`
- Las claves pueden ser strings, números o tuplas

````python
# Diccionario simple: componente -> coste
costes = {
    'A': 0.32,
    'B': 0.25,
    'C': 0.05
}

print("Coste de A:", costes['A'])
print("Coste de B:", costes['B'])

# Añadir nuevo elemento
costes['D'] = 0.40
print("\nDiccionario actualizado:", costes)
````

````python
# Diccionario con tuplas como keys
# Representa costes de transporte entre almacén y ciudad
coste_transporte = {
    ('Harlingen', 'NYC'): 1956,
    ('Harlingen', 'LA'): 1606,
    ('Memphis', 'NYC'): 1096,
    ('Memphis', 'LA'): 1792
}

# Acceder a un valor específico
print("Coste Harlingen -> NYC:", coste_transporte[('Harlingen', 'NYC')])

# También se puede escribir sin el paréntesis extra:
print("Coste Memphis -> LA:", coste_transporte['Memphis', 'LA'])
````

````python
# Iterar sobre diccionarios
print("Todas las rutas y sus costes:")
for ruta, coste in coste_transporte.items():
    almacen, ciudad = ruta  # Desempaquetar tupla
    print(f"  {almacen} -> {ciudad}: ${coste}")
````

## 4. Bucles y Comprensiones

### Bucle `for`
En Python, `for` itera directamente sobre elementos (diferente a MATLAB donde iteras sobre índices)

````python
# Iterar sobre una lista
componentes = ['A', 'B', 'C']

print("Forma Python (iterar sobre elementos):")
for comp in componentes:
    print(f"Componente: {comp}")

print("\nSi necesitas índices, usa enumerate():")
for i, comp in enumerate(componentes):
    print(f"Índice {i}: {comp}")
````

````python
# range() para generar secuencias de números
# MATLAB: for i = 1:5

print("range(5) genera: 0, 1, 2, 3, 4")
for i in range(5):
    print(i, end=' ')

print("\n\nrange(1, 6) genera: 1, 2, 3, 4, 5")
for i in range(1, 6):
    print(i, end=' ')
````

### List comprehension

Permiten crear listas de forma concisa. Muy usadas en restricciones y funciones objetivo de Pyomo.

````python
# Forma tradicional
cuadrados = []
for i in range(5):
    cuadrados.append(i**2)
print("Forma tradicional:", cuadrados)

# Comprensión de lista (más compacta)
cuadrados = [i**2 for i in range(5)]
print("Comprensión de lista:", cuadrados)
````

````python
# Ejemplos relevantes para Pyomo
componentes = ['A', 'B', 'C']
costes = {'A': 2, 'B': 2, 'C': 5}

# Crear lista de costes
lista_costes = [costes[c] for c in componentes]
print("Lista de costes:", lista_costes)

# Con filtro (if)
componentes_baratos = [c for c in componentes if costes[c] < 3]
print("Componentes baratos:", componentes_baratos)
````

### Función sum()

````python
# Sumar elementos de una lista
numeros = [1, 2, 3, 4, 5]
total = sum(numeros)
print(f"Suma de {numeros} = {total}")

# Combinar sum() con comprensión de lista (patrón típico en Pyomo)
componentes = ['A', 'B', 'C']
cantidades = {'A': 0.3, 'B': 0.5, 'C': 0.2}
costes = {'A': 2, 'B': 2, 'C': 5}

# Calcular coste total: sum(cantidad[c] * coste[c] for c in componentes)
coste_total = sum(cantidades[c] * costes[c] for c in componentes)
print(f"\nCoste total: {coste_total}")
````

## 5. Funciones

### Funciones regulares

````python
# Definir una función
def calcular_coste(cantidad, precio):
    """Calcula el coste total dado cantidad y precio."""
    return cantidad * precio

# Usar la función
coste = calcular_coste(100, 0.25)
print(f"Coste: ${coste}")
````

````python
# Función con múltiples valores de retorno
def estadisticas(numeros):
    """Retorna suma y promedio de una lista."""
    total = sum(numeros)
    promedio = total / len(numeros)
    return total, promedio  # Retorna tupla

datos = [10, 20, 30, 40]
suma, prom = estadisticas(datos)
print(f"Suma: {suma}, Promedio: {prom}")
````

### Funciones Lambda
Permiten definir restricciones simples de forma compacta

````python
# Función regular
def cuadrado(x):
    return x**2

# Función lambda equivalente (anónima, de una línea)
cuadrado_lambda = lambda x: x**2

print("Función regular:", cuadrado(5))
print("Función lambda:", cuadrado_lambda(5))
````

````python
# Lambdas con múltiples argumentos
suma = lambda a, b: a + b
print("Suma:", suma(10, 20))
````

## Resumen de Conceptos Clave para Pyomo

| Concepto | Uso en Pyomo | Ejemplo |
|----------|--------------|----------|
| Listas | Definir conjuntos | `['A', 'B', 'C']` |
| Tuplas | Índices múltiples | `(almacén, ciudad)` |
| Diccionarios | Parámetros del modelo | `{'A': 2, 'B': 3}` |
| Comprensiones | Restricciones | `[x[i] for i in Set]` |
| `sum()` | Función objetivo | `sum(c[i]*x[i] for i in Set)` |
| Lambda | Reglas simples | `lambda m: m.x + m.y == 1` |
| Bucles anidados | Dobles índices | `for i in I for j in J` |
| f-strings | Imprimir resultados | `f"x = {x.value:.2f}"` |
